# Q 7 E. Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.

```python
XT = X.T
XT
```

```python
XT.shape
```

```python
print(type(X))
```

```python
                          # should show (7, 8)

XTX = X.T @ X
print(XTX.shape)                         # (8, 8)
print(XTX)
```

```python
print(XTX)
```

# Q7 F Invert XTX.


```python
XTX_inv = np.linalg.inv(XTX)
print(XTX_inv)
print(XTX_inv.shape)
```

```python
y= np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

print(y)
print(y.shape)
```

# Q7 G Multiply the inverse of XTX with the transpose of X, and then 
# multiply the result by y. Call the result w.


```python
result=XTX_inv @ (X.T)
```

```python
w=result @ y
```

# Q7 H What is the sum of result


```python
final_sum=np.sum(w)
print(final_sum)
```
