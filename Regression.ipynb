```python
u=np.array([1,2,3,4])
```

```python
v=np.array([6,7,8,9])
```

```python
u*v 7, 9, 11,13  6, 14, 24,36
```

```python
u.shape[0]
```

```python
v.shape[0]
```

```python
def vect_mul(u,v):
    assert u.shape[0]==v.shape[0]
    n=u.shape[0]
    results=0.0
    for i in range(n):
        results+=u[i]*v[i]
    return results
    
    
```

```python
print(vect_mul(u,v))
```

```python
u1=np.array([[1,2,3,4,5],
            [5,6,7,8,5],
            [9,10,11,12,5],
            [13,14,15,16,5]])
v1=np.array([1,2,3,4,5])
```

```python
v1.shape[0] 5,10
```

```python
def vector_matrix_mul(u,v):
    assert u.shape[1]==v.shape[0]
    num_of_rows=u.shape[0]
    result=0
    for i in range(num_of_rows):
        result+=u[i]*v[i]
    return result
        
    
```

```python
print(vector_matrix_mul(u1,v1))
```
